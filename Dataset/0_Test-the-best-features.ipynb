```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import f1_score
from itertools import combinations

file_name = "D:/ZBIOT-2025/Dataset/Merged_Data/A_6-Merged_sequence_data_fix_durations/Topology_A_fix_duration.csv"

feature_columns = [
    "Average Packet Length", "Packet Standard Variance","Average IAT", "IAT Standard Variance","Sequence Number Maximum", "Sequence Number Minimum", "Sequence Number Average"]

train_data = pd.read_csv(file_name)
test_data = pd.read_csv(file_name)

train_indices = train_data.sample(frac=0.8, random_state=42).index
test_indices = train_data.drop(train_indices).index

y_train = train_data["Device Name"].iloc[train_indices].astype("category")
categories = y_train.cat.categories
y_train = y_train.cat.codes

y_test = test_data["Device Name"].iloc[test_indices].astype("category")
y_test = y_test.cat.set_categories(categories).cat.codes

class_weights = compute_class_weight(class_weight="balanced", classes=pd.Series(y_train).unique(), y=y_train)
class_weight_dict = {i: weight for i, weight in enumerate(class_weights)}

results = []
for r in range(1, len(feature_columns)+1):
    for subset in combinations(feature_columns, r):
        selected_features = list(subset)

        X_train = train_data.loc[train_indices, selected_features]
        X_test = test_data.loc[test_indices, selected_features]

        model = RandomForestClassifier(random_state=42, class_weight=class_weight_dict)
        model.fit(X_train, y_train)
        y_pred = model.predict(X_test)
        f1 = f1_score(y_test, y_pred, average='macro')

        results.append((selected_features, f1))
        print(f"Features: {selected_features} → F1-score: {f1:.4f}")

sorted_results = sorted(results, key=lambda x: x[1], reverse=True)
for i, (features, score) in enumerate(sorted_results[:5], 1):
    print(f"{i}. {features} → F1-score: {score:.4f}")

```
