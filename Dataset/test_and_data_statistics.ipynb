```python
import pandas as pd
import os
from glob import glob

def search_entry_in_files(file_path, search_term, column_name='Device Name', filter_keyword='groundtruth'):
    """
    Search for a specific entry in CSV files and count occurrences in each file.
    
    Parameters:
    -----------
    file_path : str
        Path to the directory containing CSV files
    search_term : str
        The entry to search for (e.g., device name)
    column_name : str, optional
        Column name to search in (default: 'Device Name')
    filter_keyword : str, optional
        Keyword to filter files (default: 'groundtruth')
    
    Returns:
    --------
    dict
        Dictionary with file names as keys and occurrence counts as values
    pd.DataFrame
        Summary DataFrame with columns: ['File', 'Occurrences', 'Total_Rows']
    """
    
    # Dictionary to store results
    results = {}
    
    # Find all CSV files in the directory and subdirectories
    csv_files = glob(os.path.join(file_path, '**', '*.csv'), recursive=True)
    
    # Iterate over CSV files
    for file in csv_files:
        # Apply filter if specified
        if filter_keyword and filter_keyword not in file:
            continue
        
        try:
            # Read the CSV file
            df = pd.read_csv(file)
            
            # Check if the column exists
            if column_name not in df.columns:
                print(f"Warning: Column '{column_name}' not found in {file}")
                continue
            
            # Count occurrences of the search term
            count = (df[column_name] == search_term).sum()
            
            # Store results
            file_name = os.path.basename(file)
            results[file_name] = {
                'occurrences': count,
                'total_rows': len(df),
                'full_path': file
            }
            
            if count > 0:
                print(f"Found {count} occurrence(s) in: {file}")
        
        except Exception as e:
            print(f"Error reading {file}: {str(e)}")
    
    # Create summary DataFrame
    summary_df = pd.DataFrame([
        {
            'File': file_name,
            'Occurrences': data['occurrences'],
            'Total_Rows': data['total_rows'],
            'Percentage': (data['occurrences'] / data['total_rows'] * 100) if data['total_rows'] > 0 else 0,
            'Full_Path': data['full_path']
        }
        for file_name, data in results.items()
    ])
    
    # Sort by occurrences (descending)
    if not summary_df.empty:
        summary_df = summary_df.sort_values('Occurrences', ascending=False).reset_index(drop=True)
    
    return results, summary_df


# Example usage:
if __name__ == "__main__":
    file_path = './Data/'
    search_term = 'Sonoff Motion 2'  # Replace with actual device name
    
    # Search for the entry
    results_dict, summary_df = search_entry_in_files(
        file_path=file_path,
        search_term=search_term,
        column_name='Device Name',
        filter_keyword='groundtruth'
    )
    
    # Display summary
    print("\n=== SUMMARY ===")
    print(summary_df)
    
    # Total occurrences across all files
    total_occurrences = summary_df['Occurrences'].sum()
    print(f"\nTotal occurrences across all files: {total_occurrences}")
    
    # Files with most occurrences
    if not summary_df.empty:
        print(f"\nFile with most occurrences: {summary_df.iloc[0]['File']} ({summary_df.iloc[0]['Occurrences']} times)")
```

```python
# Remove nan entries in 'Source Address ZigBee' column
data_zbee_r = data.dropna(subset=['Source Address ZigBee'])

# remove entries where Source Address do not coincide with Source Address ZigBee
data_zbee = data_zbee_r[data_zbee_r['Source Address'] == data_zbee_r['Source Address ZigBee']]

len(data_zbee), len(data_zbee_r)
```

```python
columns = ['Device Name ZigBee', 'Device Name ZigBee Destination']
combined_results = {}

for col in columns:
    if  col == 'Device Name ZigBee Destination':
        #filter only for messages which have as destination a coord, unknown or broadcast
        data = data_zbee[data_zbee['Device Name ZigBee'].isin(['Coordinator', 'Unknown', 'Broadcast'])]
    else:
        data = data_zbee
    
    value_counts = data[col].value_counts()
    human_command_counts = data[data['Human Command'] == 1][col].value_counts()
    
    for value, count in value_counts.items():
        human_command_count = human_command_counts.get(value, 0)
        
        if value not in combined_results:
            combined_results[value] = {'total_count': 0, 'human_command_count': 0}
        
        combined_results[value]['total_count'] += count
        combined_results[value]['human_command_count'] += human_command_count

# Print merged results
for value, counts in sorted(combined_results.items(), key=lambda x: x[1]['total_count'], reverse=True):
    total = counts['total_count']
    human_cmd = counts['human_command_count']
    percentage = (human_cmd / total) * 100 if total > 0 else 0
    percentage_count = (total / len(data_zbee)) * 100 if len(data_zbee) > 0 else 0
    print(f'Value: {value}, Count: {total}, Human Command Count: {human_cmd}, Percentage: {percentage:.2f}%, Percentage of Total Messages: {percentage_count:.2f}%')

print(f'Total messages: {sum(c["total_count"] for c in combined_results.values())}')
print(f'Total human commands: {sum(c["human_command_count"] for c in combined_results.values())}')

print(len(data_zbee))
```

```python
import csv
import re

# Prepare data for CSV
csv_data = []
for value, counts in sorted(combined_results.items(), key=lambda x: x[1]['total_count'], reverse=True):
    total = counts['total_count']
    human_cmd = counts['human_command_count']
    percentage = (human_cmd / total) * 100 if total > 0 else 0
    percentage_count = (total / len(data_zbee)) * 100 if len(data_zbee) > 0 else 0
    
    csv_data.append({
        'Device': value,
        'Total_Count': total,
        'Human_Command_Count': human_cmd,
        'Percentage': round(percentage, 2),
        'Percentage_of_Total_Messages': round(percentage_count, 2)
    })

# Convert to DataFrame and save as CSV
results_df = pd.DataFrame(csv_data)
# results_df.to_csv('combined_zigbee_results.csv', index=False)

# Group devices by name pattern (merge similar names differing only in last character)

merged_results = {}

for device, data in results_df.iterrows():
    device_name = data['Device']
    # Remove trailing numbers/characters to group similar devices
    base_name = re.sub(r'[\s\d]+$', '', device_name)
    
    if base_name not in merged_results:
        merged_results[base_name] = {
            'devices': [],
            'total_percentage': 0
        }
    
    merged_results[base_name]['devices'].append(device_name)
    merged_results[base_name]['total_percentage'] += data['Percentage_of_Total_Messages']

# Calculate averages and display
merged_df_list = []
for base_name, data in sorted(merged_results.items(), key=lambda x: x[1]['total_percentage'], reverse=True):
    avg_percentage = data['total_percentage'] / len(data['devices'])
    merged_df_list.append({
        'Device_Group': base_name,
        'Count': len(data['devices']),
        'Devices': ', '.join(data['devices']),
        'Average_Percentage_of_Total': round(avg_percentage, 2)
    })

merged_df = pd.DataFrame(merged_df_list)
merged_df.to_csv('merged_zigbee_results.csv', index=False)
print(merged_df)
```
